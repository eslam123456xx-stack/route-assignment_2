```python
#problem 2
def area(redius,shape):
    if shape=="circle":
        return 3.14*redius**2
    elif shape=="square":
        return redius**2
    elif shape=="rectangle":
        return redius*2
    else:
        return "Invalid shape"
print(area(5,"circle"))
print(area(5,"square"))
```

```python
#problem 5
class resturant:
    def __init__(self):
        self.menu_items = {}
        self.book_tables =[]
        self.customer_order={}


    def add_menu_item(self, item_name, item_price):
        self.menu_items[item_name] = item_price
        print(f"{item_name} added to the menu with price {item_price}")


    def book_table(self, table_number):
        self.book_tables.append(table_number)
        print(f"Table {table_number} booked successfully") 


    def take_order(self, table_number, order):
        if table_number in self.book_tables:
            self.customer_order[table_number] = order
            print(f"Order taken for table {table_number}: {order}")
        else:
            print(f"Table {table_number} is not booked. Please book the table first.")


    def display_menu(self):
        print("Menu:")
        for item_name, item_price in self.menu_items.items():
            print(f"{item_name}: {item_price}")

    def display_booked_tables(self):
        print("Booked Tables:")
        for table_number in self.book_tables:
            print(f"Table {table_number}")

    def display_customer_orders(self):
        print("Customer Orders:")
        for table_number, order in self.customer_order.items():
            print(f"Table {table_number}: {order}")

resturant = resturant()
resturant.add_menu_item("Pizza", 10)
resturant.add_menu_item("Burger", 5)
resturant.book_table(1)
resturant.take_order(2, "Pizza")
resturant.take_order(1, "Pizza")
resturant.display_menu()
resturant.display_booked_tables()
resturant.display_customer_orders()            
```
